```python
import pandas as pd

id_cols = ['flight_id', 'timestamp']
df = pd.read_parquet("EGLL_dataset.parquet.gz",columns=id_cols)
df
```

```python
groups = df.groupby('flight_id',sort=False)

groups.describe()
```

```python
min_vals = groups.min()
max_vals = groups.max()
min_vals['flight_id_col'] = min_vals.index
max_vals['flight_id_col'] = max_vals.index
```

```python
day_grouper = pd.Grouper(key='timestamp', freq='D')
start_groups = min_vals.groupby(day_grouper,sort=True)
end_groups = max_vals.groupby(day_grouper,sort=True)
```

```python
i = 0
for key in start_groups.groups.keys():
    starting = start_groups.get_group(key)
    ending = end_groups.get_group(key)
    interval = starting.merge(ending, on='flight_id', how='outer', suffixes=('_start', '_end')).drop(columns=['flight_id_col_start', 'flight_id_col_end'])
    print(interval)
    i += 1
    
    daydata = pd.read_parquet("data/EGLL_dataset.parquet.gz",filters=[('flight_id', 'in', interval.index)])
    print(daydata)
    if i > 5:
        break
```
